# Data Storytelling Example: Quality of Life Scores
## Motivation
Let's say that we are interested in how quality of life varies across different urban areas. Some reasons we might be interested in this include:
* Moving to a new area and wanting to do some research beforehand
* Considering establishing a new business and want to make sure the area can support it
* Etc. 

There are several aspects of quality of life that can be used to determine good places to live or start businesses. For example:
* Safety/crime rate
* Education
* Cost of living
* Access to outdoors, parks, recreation
* Traffic/travel connectivity, public transportation
* Etc. 

Originally, we could get this information from the [Teleport API](https://developers.teleport.org/). Unfortunately, this API is no longer freely available for us to use. Instead, we will use a saved CSV dataset containing Seattle's quality of life scores from the original example.

**Note:** These are historical scores, not current measurements. The original dataset did not include a date for when the scores were collected. We are using them to practice data storytelling, not to make up-to-date comparisons between cities.

## Loading the Data
Let's load the quality of life data for Seattle. Seattle will be our *proof of concept*, showing that we can get the data we want and use it to tell a story.

Instead of making an API request, we will load the scores from a saved CSV dataset. The data is included directly in this notebook so it runs without an internet connection or API key. You can also load the accompanying `seattle_quality_of_life_scores.csv` file using `pd.read_csv("seattle_quality_of_life_scores.csv")`.

```python
import pandas as pd

scores_df = pd.read_csv("seattle_quality_of_life_scores.csv")
print(scores_df)
```

## Organizing the Data
Now that we have the data for Seattle, we want to get the category and score information out of it. We will store these category:score pairs in a pandas series for easy data exploration later.

```python
cat_score_ser = scores_df.set_index("name")["score_out_of_10"]
print(cat_score_ser)
```

## Data Visualization
Let's visualize our Seattle quality of life score data using a bar chart (x-axis will be categories and y-axis will be scores).

```python
import matplotlib.pyplot as plt
# GS: answering question after class: add code to change labels?
# color = 'blue'
# plt.rcParams['text.color'] = color
# plt.rcParams['axes.labelcolor'] = color
# plt.rcParams['xtick.color'] = color
# plt.rcParams['ytick.color'] = color

plt.bar(cat_score_ser.index, cat_score_ser)
plt.xticks(rotation=45, ha="right")
plt.ylabel("Score out of 10")
plt.title("Seattle Quality of Life Scores")
plt.show()
```

## Insights
Some strengths of Seattle include:
1. Startups
1. Business freedom

Some weaknesses of Seattle include:
